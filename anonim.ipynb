# Firefox

```python
from selenium import webdriver
from selenium.webdriver.firefox.options import Options
from selenium.webdriver.firefox.firefox_profile import FirefoxProfile
import time

profile = FirefoxProfile()
profile.set_preference("general.useragent.override", "Godzilla/5.0 (Linux Ubuntu 22.04 LTS; x64; rv:22.04) Gecko/20100101 Firebox/121.0")

options = Options()
options.profile = profile

driver = webdriver.Firefox(options=options)
driver.get('https://whatmyuseragent.com/')
time.sleep(10)
driver.quit()
```

```python
#223.205.29.143:8080  Тайвань https
profile = FirefoxProfile()

# Ручная настройка прокси
profile.set_preference("network.proxy.type", 1)  # 1 = ручная настройка
profile.set_preference("network.proxy.http", "223.205.29.143")
profile.set_preference("network.proxy.http_port", 8080)
profile.set_preference("network.proxy.ssl", "223.205.29.143")
profile.set_preference("network.proxy.ssl_port", 8080)
profile.set_preference("network.proxy.no_proxies_on", "localhost, 127.0.0.1")

options = Options()
options.profile = profile

driver = webdriver.Firefox(options=options)
driver.get('https://whatmyuseragent.com/')
time.sleep(10)
driver.quit()
```
